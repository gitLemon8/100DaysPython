## **Python Indentation**
Indentation -> spaces at the beginning of a code line.
Python uses indentation to indicate a block of code.
Example

````python
if 5 > 2:
  print("Five is greater than two!")
````

Python will give us an error, specifically indentation error if we miss the indentation. Example:

````python
if 5 > 2:
  print("Five is greater than two!")
````

The number of spaces is up to us as a programmer, the most common use is four, but it has to be at least one. Some code editor already add tab as "four spaces" so we can just use that.
Example:

````python
if 5 > 2:
  print("Five is greater than two!");
if 5 > 2:
        print("Five is greater than two!")
````

Both still work. We have to use the same number of spaces in the same block of code, otherwise Python will give us an error:

````python
if 5 > 2:
  print("Five is greater than two");
    print("Five is greater than two");
````

# **Python Output / `print()`**
You can use the `print()` to display text or output values.

````python
print("Hello World!")
````

You can use the `print()` function as many times as you want. Each call prints text on a new line by default

````python
print("Hello World!")
print("I am learning Python.")
print("It is nice")
````

Text in Python must be inside quotes. You can use either `"` double quotes or `'` single quotes

````python
print("This work")
print("This also work")
````

## Print Without a New Line
By default, the print() function ends with a new line. If you want to print multiple words on the same line, you can use the `end` parameter

````python
print("Hello World!", end=" ")
print("This will print on the same line.")
````

## Print Numbers
We can also use the `print()` function to display numbers. However, unlike text we don't put numbers inside quotes:

````python
print(3)
print(358)
print(50000)
````

It is also possible to do math inside print() function

````python
print(3 + 3)
print(2 * 5)
````

You can combine text and numbers in one output by separating them with a comma

````python
print("I am", 35, "years old.")
````
